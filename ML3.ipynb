```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import accuracy_score, classification_report, f1_score
```

```python
baca = pd.read_csv("go_track_tracks.csv")
baca.head()
```

```python
baca.info()
```

```python
baca = baca.drop(["linha"], axis=1)
baca.head()

```

```python
#Menentukan variabel yang akan di klusterkan
baca_x = baca.iloc[:, 1:3]
baca_x.head()

	#Memvisualkan persebaran data
plt.scatter(baca.distance, baca.speed, s =10, c = "red", marker = "o", alpha = 0.5)
plt.show()
```

```python
x_array = np.array(baca_x)
print(x_array)

```

```python
scaler = MinMaxScaler()
x_scaled = scaler.fit_transform(x_array)
x_scaled

```

```python
from sklearn.cluster import KMeans

# Membuat model KMeans
kmeans = KMeans(n_clusters=3, random_state=42)

# Training model
kmeans.fit(x_scaled)

# Menampilkan label cluster
print(kmeans.labels_)

# Menambahkan hasil cluster ke dataframe
baca["kluster"] = kmeans.labels_

# Menampilkan data
print(baca.head())
```

```python
y_test = baca["kluster"]
y_pred = kmeans.labels_

# %%
# Accuracy
print("Accuracy Score:", accuracy_score(y_test, y_pred))

# %%
# Classification report
print(f'Classification Report: \n{classification_report(y_test, y_pred)}')

# %%
# F1 score
print(f"F1 Score : {f1_score(y_test, y_pred, average='weighted')}")
```

```python
import matplotlib.pyplot as plt

# Plot hasil clustering
plt.figure(figsize=(8,6))

# Memvisualisasikan sebaran data berdasarkan kluster
plt.scatter(
    baca['id_android'],
    baca['speed'],
    c=baca['kluster'],
    cmap='viridis',
    alpha=0.6
)

# Titik centroid (Diubah menjadi indeks 0 dan 1)
plt.scatter(
    kmeans.cluster_centers_[:, 0],   # Sumbu X (Fitur pertama)
    kmeans.cluster_centers_[:, 1],   # Sumbu Y (Fitur kedua)
    s=200,
    c='red',
    marker='X',
    label='Centroid'
)

# Sesuaikan label dengan kolom yang di-plot
plt.xlabel('ID Android')
plt.ylabel('Speed')
plt.title('K-Means Clustering')
plt.legend()

plt.show()
```
